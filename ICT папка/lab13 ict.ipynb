```python
import matplotlib.pyplot as plt
x = [2, 4, 8, 12, 18]
y = [3, 9, 2, 8, 14]
# 1. --' үзік сызық
plt.subplot(2, 2, 1)
plt.plot(x, y, '--')
plt.title('Үзік сызық')
# 2. '-r' қызыл тұтас сызық
plt.subplot(2, 2, 2)
plt.plot(x, y, '-r')
plt.title('Қызыл сызық')
# 3. 'ro' қызыл дөңгелек нүкте
plt.subplot(2, 2, 3)
plt.plot(x, y, 'ro')
plt.title('Нүктелі график')
# 4. 'bx' көк x маркер
plt.subplot(2, 2, 4)
plt.plot(x, y, 'bx')
plt.title('X маркер')
plt.suptitle('Сызық түрлері', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [2, 4, 8, 12, 18]
y1 = [3, 9, 2, 8, 14]
y2 = [5, 3, 11, 6, 9]
y3 = [7, 2, 9, 4, 13]
y4 = [1, 6, 4, 11, 8]

plt.subplot(2, 2, 1)
plt.plot(x, y1, '-o', color='red', linewidth=2, markersize=6)
plt.title('График 1')

plt.subplot(2, 2, 2)
plt.plot(x, y2, '--s', color='green', linewidth=2, markersize=6)
plt.title('График 2')

plt.subplot(2, 2, 3)
plt.plot(x, y3, '-^', color='blue', linewidth=2, markersize=6)
plt.title('График 3')

plt.subplot(2, 2, 4)
plt.plot(x, y4, ':D', color='purple', linewidth=2, markersize=6)
plt.title('График 4')

plt.suptitle('Менің графиктерім', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
# Деректер
destinations = ["Дубай", "Стамбул", "Бангкок", "Париж", "Рим", "Бали", "Барселона"]
tourists = [1200, 980, 850, 1500, 760, 920, 680]
# DataFrame құру
df = pd.DataFrame({
    "Қала": destinations,
    "Туристер саны": tourists
})
# Кестені шығару
print(df)
# Диаграмма салу
plt.figure()
bars = plt.bar(df["Қала"], df["Туристер саны"], color='red')
# Бағандардың үстіне сан жазу
for bar in bars:
    plt.text(bar.get_x() + bar.get_width()/2,
             bar.get_height() + 10,
             str(bar.get_height()),
             ha='center', fontsize=9)
# Атаулар
plt.title("Танымал саяхат бағыттары (туристер саны)")
plt.xlabel("Қалалар")
plt.ylabel("Туристер саны")
plt.xticks(rotation=15)
# Көрсету
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
# Деректер
years = ["2018", "2019", "2020", "2021", "2022", "2023", "2024"]
bachelor = [1200, 1350, 1100, 1450, 1600, 1750, 1900]
master = [320, 380, 290, 410, 450, 520, 580]
# DataFrame құру
df = pd.DataFrame({
    "Жыл": years,
    "Бакалавр": bachelor,
    "Магистратура": master
})
print("Нархоз университетіне түскен студенттер кестесі:")
display(df)
# Диаграмма салу
plt.figure(figsize=(14, 8))
bar_width = 0.35
r = np.arange(len(df["Жыл"]))
# Бакалавр бағандары
plt.bar(r - bar_width/2, df["Бакалавр"], color='skyblue', width=bar_width, label='Бакалавр')
# Магистратура бағандары
plt.bar(r + bar_width/2, df["Магистратура"], color='lightcoral', width=bar_width, label='Магистратура')
# Атаулар
plt.title('Нархоз университетіне жыл сайын түсетін студенттер саны', fontsize=16)
plt.xlabel('Жыл', fontsize=12)
plt.ylabel('Студенттер саны', fontsize=12)
plt.xticks(r, df["Жыл"], rotation=45, ha='right')
# Сандарды бағандар үстіне жазу
for i in range(len(df["Жыл"])):
    plt.text(r[i] - bar_width/2, df["Бакалавр"][i], f'{df["Бакалавр"][i]:.0f}', ha='center', va='bottom', fontsize=9)
    plt.text(r[i] + bar_width/2, df["Магистратура"][i], f'{df["Магистратура"][i]:.0f}', ha='center', va='bottom', fontsize=9)
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
# Деректер
faculties = ["Бизнес", "Экономика", "IT", "Құқық", "Қаржы", "Маркетинг", "Туризм"]
students = [850, 720, 650, 480, 920, 390, 310]
# DataFrame құру
df_pie_chart = pd.DataFrame({
    "Факультет": faculties,
    "Студенттер": students
})
display(df_pie_chart)
# Пайыз + абсолютті сан функциясы
def func(pct, allvals):
    absolute = int(pct/100.*np.sum(allvals))
    return f"{pct:.1f}%\n({absolute:d})"
plt.figure(figsize=(10, 8))
wedges, texts, autotexts = plt.pie(
    df_pie_chart["Студенттер"],
    labels=df_pie_chart["Факультет"],
    autopct=lambda pct: func(pct, df_pie_chart["Студенттер"]),
    startangle=90,
    pctdistance=0.85,
    colors=plt.cm.Paired.colors
)
plt.title("Нархоз университетіндегі факультеттер бойынша студенттер үлесі", fontsize=16)
plt.axis("equal")
plt.setp(autotexts, size=10, weight="bold")
plt.setp(texts, size=11)

plt.show()
```
